### Helper function to collect company's ratios.

```python
def collect_ratios(tickers: list, ratios: list):
    rows = []

    for ticker in tickers:
        info = yf.Ticker(ticker).info
        row = [ticker] + [info.get(ratio, None) for ratio in ratios]
        rows.append(row)

    return pd.DataFrame(rows, columns=["Ticker"] + ratios)
```

### Helper function to collect specific ratios

```python
def get_financial_ratios(company):
    """
    Calculates key financial ratios from a yfinance Ticker object.
    """
    try:
        # 1. Identify the most recent reporting column
        # yfinance colums are usually timestamps (latest is at index 0)
        col_inc = company.income_stmt.columns[0]
        col_bs = company.balance_sheet.columns[0]

        # 2. Extract specific line items
        # Using .loc to grab the exact row and the latest date column
        net_income = company.income_stmt.loc["Net Income", col_inc]
        total_revenue = company.income_stmt.loc["Total Revenue", col_inc]
        total_assets = company.balance_sheet.loc["Total Assets", col_bs]
        equity = company.balance_sheet.loc["Stockholders Equity", col_bs]

        # Handle 'Total Debt' (sometimes missing or named differently)
        if "Total Debt" in company.balance_sheet.index:
            total_debt = company.balance_sheet.loc["Total Debt", col_bs]
        else:
            total_debt = 0  # Fallback if no debt is found

        # 3. Ratio Calculations
        ratios = {
            "Return on Assets (ROA) %": (net_income / total_assets) * 100,
            "Return on Equity (ROE) %": (net_income / equity) * 100,
            "Profit Margin %": (net_income / total_revenue) * 100,
            "Asset Turnover Ratio": total_revenue / total_assets,
            "Debt-to-Equity": total_debt / equity
        }

        return ratios

    except KeyError as e:
        return f"Error: The metric {e} was not found in the financial statements."
    except Exception as e:
        return f"An unexpected error ocurred: {e}"
```

```python
# --- Example Usage ---
ticker_symbol = "MSFT"
company = yf.Ticker(ticker_symbol)
results = get_financial_ratios(company)

if isinstance(results, dict):
    print(f"--- Financial Ratios for {ticker_symbol} ---")
    for ratio, value in results.items():
        print(f"{ratio}: {value:.4f}")
else:
    print(results)
```

### Debt-to-Equity Visualizer

```python
def plot_debt_to_equity(ticker_symbol, years=4):
    """
    Plots the Debt-to-Equity ratio for a given ticker over a specific number of years.
    """
    try:
        # 1. Fetch data
        company = yf.Ticker(ticker_symbol)
        bs = company.balance_sheet
        
        # Check if we have enough data columns
        available_years = bs.shape[1]
        num_years = min(years, available_years)
        
        # 2. Extract and reverse data (so it plots from oldest to newest)
        # We slice the first 'num_years' and then reverse with [::-1]
        subset_bs = bs.iloc[:, :num_years][::-1]
        
        dates = subset_bs.columns
        ratios = []
        
        for date in dates:
            total_debt = subset_bs.loc["Total Debt", date] if "Total Debt" in subset_bs.index else 0
            equity = subset_bs.loc["Stockholders Equity", date]
            
            # Calculate ratio
            de_ratio = total_debt / equity
            ratios.append(de_ratio)
            
        # 3. Create the plot
        plt.figure(figsize=(10, 6))
        plt.plot(dates.strftime('%Y-%m-%d'), ratios, marker='o', linestyle='-', color='b', linewidth=2)
        
        # Formatting
        plt.title(f"Debt-to-Equity Ratio Trend: {ticker_symbol.upper()}", fontsize=14)
        plt.xlabel("Fiscal Year End", fontsize=12)
        plt.ylabel("Ratio (Debt/Equity)", fontsize=12)
        plt.grid(True, linestyle='--', alpha=0.7)
        
        # Add values as text on the points
        for i, val in enumerate(ratios):
            plt.text(i, val, f'{val:.2f}', ha='center', va='bottom', fontsize=10)

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"Error analyzing {ticker_symbol}: {e}")
```

```python
plot_debt_to_equity("AAPL", years=5)
```
